# What is AI? What is machine learning? What is an LLM (large language model)?

Artificial intelligence has been around since the 1950s. Even in the early 1990s, AI had a lot of theoretical promise, but very little to show for all of the research that had been done.

Machine learning is a new(ish) technique in the AI space. It basically says: Don't try to write all of the rules that a computer should follow to think like a human. Rather, show it lots of examples and let it extrapolate from those.   Modern machine learning is all about training a model with as much input as possible, so that when we ask it to categorize data, it can do that as well as possible.  The more data you have, and the better a model you have, the more accurate the AI will be.

People started to talk about "generative AI," that instead of asking the AI to recognize and predict words, we would use it to generate text. In other words: We would ask it to start a sentence with "the," then ask it to generate the most likely word to come after "the". Then the most likely word to come after that. Add some statitics and variance, and give it a "hint" as to the topic it should write about, and we can generate text that approximates what humans do.

When ChatGPT came out two years ago, it was not the first, and not the only, generative AI system, but it was clearly ahead of the rest. Now we have dozens? hundreds? of startups (and big companies) all racing to create machine-learning systems that can generate text, in order to answer questions that we pose. It's really just combining all of the text that we have trained it on, and then producing the best, most likely result that we want.

We can have it write text, write poetry, write code... It won't necessarily create the *best* text or code or poetry, because it's basically giving us the average of everything it was ever trained on. But that's still good enough for a lot of what we want to do.

# Choosing a model

When we use a chatbot / large-language model, this means that we have to choose from among a bunch of existing models. Each model is different in the way that it responds, the depth to which it "thinks," and also the speed with which it responds.

# Can it create code? Should we use it to create code?

These models can indeed create code, and their code is usually good. However, chatbots are known to "hallucinate," meaning that they come up with "facts" that aren't true, and code that doesn't work.

Sometimes the code won't work because it has bugs. Often, the code won't work because it references functions and modules that don't exist.

English and other human languages are ambiguous. When we say something, it can be interpreted in many different ways. Programming is all about being clear, unambiguous, and specific. There's always going to be a mismatch.  

Assume that the code you will get from a chatbot is wrong and/or not as good as it could be.



# If I'm not going to have it generate code, what are doing here?

I love to *argue* with LLMs. I don't accept what they say at face value, but I challenge them. I ask them lots of questions. Only after getting lots of answers to lots of questions am I satisfied that they know what they're talking about. (Often, they'll change their tune.)  You have to argue with it.

I call this the "reverse Socratic method" of learning. After asking the LLM lots of questions, *I* have learned a lot about the subject.

# Pair programming 

You can (obviously) write code on your own. But it's often better to write code along with someone else, two people on the same computer (or two people on their own computers sharing editing abilities, a la Google Docs). When you are forced to speak with someone else about your programming, you are forced to think it through and crystallize your plans. That can expose flaws and issues.

I am a big fan of pair programming with an LLM. This does *not* mean that I ask the LLM to write the code, and I review it. Rather, it means that I use the LLM as a sounding board for my plans, ideas, code, and even for doing some of the stuff that I don't want to.

How, then, can we use an LLM for pair programming if not for generating code?

1. Talking through your strategy.
2. Reviewing your code after you've written it.
3. Debugging and/or working through issues/problems/roadblocks.

# Talking through the strategy

Use the LLM as a pair, or even as a "rubber duck."  Before writing any code, tell the LLM how you plan to solve it. Get feedback. Ask it what the good and bad parts are.  You can, in your text prompt to the LLM, say what you're looking to do. Try to get it *NOT* to generate code.

Then.. challenge it.

# Example: Counting vowels

I want to write a Python function, `count_vowels`, that will take a string as input, and will return an integer as output, the number of vowels in the string.

```python
def count_vowels(text):
    all_vowels = set('aeiou')
    total = 0

    for one_character in text.lower():
        if one_character in all_vowels:
            total += 1

    return total
```

```python
count_vowels('hello out there')
```

# Exercise: Pig Latin translator

Pig Latin is a children's "secret" language, in which we can translate a word from English into Pig Latin using the following rules:

- If the first letter is a vowel (a e i o u) then we add `way` to the word.
- Otherwise, we move the first letter to the end of the word, and add `ay`.

Examples:

- `elephant` -> `elephantway`
- `apple` -> `appleway`
- `computer` -> `omputercay`
- `papaya` -> `apayapay`

Write a function, `plword`, that does this translation. Assume that the word is all lowercase, has no spaces, has no punctuation.

Use an LLM to critique your strategy and then any code you write.

```python
def plword(word):
    if word[0] in 'aeiou':
        return word + 'way'
    else:
        return word[1:] + word[0] + 'ay'
```

# Exercise: Ubbi Dubbi

Ubbi Dubbi is another secret children's language. Here, you take any vowel (a, e, i, o, u) and you put the syllable `ub` before it. 

- `milk` -> `mubilk`
- `hello` -> `hubellubo`

Write a function, `ubbi_word` that (like before) takes a single word in English -- all lowercase, no spaces, no punctuation --  and returns a string with the word translated into Ubbi Dubbi.

Use Claude to (a) run your strategy by it and (b) comment on your implementation.

```python
def ubbi_word(word):
    output = []

    for one_character in word:
        if one_character in 'aeiou':
            output.append('ub')
        output.append(one_character)

    return ''.join(output)
```

```python
ubbi_word('hello')
```

```python
ubbi_word('python')
```

# Testing

Modern software needs to be tested. In Python, we often use the `pytest` package to test our code. We write our code in a file (`prog.py`) and then put tests in a file alongside it (`test_prog.py`) and then run the tests by running `pytest` in that directory.

There are a few different approaches to testing with LLMs:

- Have the LLM generate the tests, because this is annoying and boilerplate work. This is a starting point, not an ending point, if you use this strategy!
- Have the LLM evaluate where there are holes in your tests
- Write the tests yourself, and maybe have the LLM give suggestions
